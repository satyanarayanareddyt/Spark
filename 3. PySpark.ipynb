###### String Functions

```python
# =============================================
# STRING FUNCTIONS — Complete Training Demo
# =============================================

from pyspark.sql.functions import (
    col, lit, lower, upper, initcap,
    trim, ltrim, rtrim,
    lpad, rpad,
    length, concat, concat_ws,
    substring, split, regexp_extract, regexp_replace,
    instr, locate,
    reverse, repeat,
    translate, overlay,
    when
)

# --- Sample DataFrame ---
df = spark.createDataFrame(
    [
        (1, "  Hello World  ", "alice@gmail.com", "2026-04-30"),
        (2, "APACHE SPARK", "bob@outlook.com", "2025-06-15"),
        (3, "pySpark is GREAT", "charlie@yahoo.com", "2024-12-25"),
        (4, "john doe", "diana@gmail.com", "2026-01-01"),
        (5, None, None, None),
    ],
    ["id", "text", "email", "date_str"]
)

print("=== Original Data ===")
display(df)

# =============================================
# 1. LOWER, UPPER, INITCAP — Case conversion
# =============================================

print("=== lower(), upper(), initcap() ===")
df_case = (
    df
    .withColumn("lower", lower(col("text")))          # "hello world"
    .withColumn("upper", upper(col("text")))          # "HELLO WORLD"
    .withColumn("initcap", initcap(col("text")))      # "Hello World"
)
display(df_case)

# =============================================
# 2. TRIM, LTRIM, RTRIM — Remove whitespace
# =============================================

print("=== trim(), ltrim(), rtrim() ===")
df_trim = (
    df
    .withColumn("trimmed", trim(col("text")))          # "Hello World"
    .withColumn("left_trimmed", ltrim(col("text")))    # "Hello World  "
    .withColumn("right_trimmed", rtrim(col("text")))   # "  Hello World"
)
display(df_trim)

# =============================================
# 3. LENGTH — String length
# =============================================

print("=== length() ===")
df_len = df.withColumn("text_length", length(col("text")))
display(df_len)

# =============================================
# 4. CONCAT & CONCAT_WS — Join strings
# =============================================

print("=== concat() & concat_ws() ===")
df_concat = (
    df
    .withColumn("concat", concat(lit("ID="), col("id").cast("string"), lit(" | "), col("text")))
    .withColumn("concat_ws", concat_ws(" - ", col("id").cast("string"), col("text"), col("email")))
)
display(df_concat)
# concat: no separator control, just joins
# concat_ws: joins with a separator (" - ")

# =============================================
# 5. SUBSTRING — Extract part of string
# =============================================

print("=== substring() ===")
df_sub = (
    df
    .withColumn("first_5", substring(col("text"), 1, 5))        # First 5 chars (1-based index)
    .withColumn("from_3_len_4", substring(col("text"), 3, 4))   # 4 chars starting at position 3
)
display(df_sub)

# =============================================
# 6. SPLIT — Split string into array
# =============================================

print("=== split() ===")
df_split = (
    df
    .withColumn("words", split(trim(col("text")), " "))
    .withColumn("first_word", split(trim(col("text")), " ")[0])
    .withColumn("email_domain", split(col("email"), "@")[1])
)
display(df_split)
# "Hello World" → ["Hello", "World"]
# "alice@gmail.com" → "gmail.com"

# =============================================
# 7. INSTR & LOCATE — Find position of substring
# =============================================

print("=== instr() & locate() ===")
df_pos = (
    df
    .withColumn("instr_@", instr(col("email"), "@"))       # Position of @ (1-based, 0 if not found)
    .withColumn("locate_@", locate("@", col("email")))     # Same as instr but arg order is different
)
display(df_pos)

# =============================================
# 8. LPAD & RPAD — Pad strings
# =============================================

print("=== lpad() & rpad() ===")
df_pad = (
    df
    .withColumn("id_padded", lpad(col("id").cast("string"), 5, "0"))    # "00001"
    .withColumn("text_rpad", rpad(col("text"), 25, "."))                # "Hello World.........."
)
display(df_pad)

# =============================================
# 9. REGEXP_REPLACE — Replace using regex
# =============================================

print("=== regexp_replace() ===")
df_regex_r = (
    df
    .withColumn("no_spaces", regexp_replace(col("text"), " ", ""))                # Remove all spaces
    .withColumn("mask_email", regexp_replace(col("email"), "(.*)@", "***@"))      # Mask email username
)
display(df_regex_r)

# =============================================
# 10. REGEXP_EXTRACT — Extract using regex
# =============================================

print("=== regexp_extract() ===")
df_regex_e = (
    df
    .withColumn("email_user", regexp_extract(col("email"), r"^(.+)@", 1))       # "alice"
    .withColumn("email_domain", regexp_extract(col("email"), r"@(.+)\.", 1))    # "gmail"
)
display(df_regex_e)

# =============================================
# 11. TRANSLATE — Character-level replacement
# =============================================

print("=== translate() ===")
df_trans = df.withColumn("translated", translate(col("text"), "aeiou", "12345"))
display(df_trans)
# a→1, e→2, i→3, o→4, u→5
# "Hello World" → "H2ll4 W4rld"

# =============================================
# 12. REVERSE — Reverse a string
# =============================================

print("=== reverse() ===")
df_rev = df.withColumn("reversed", reverse(trim(col("text"))))
display(df_rev)
# "Hello World" → "dlroW olleH"

# =============================================
# 13. REPEAT — Repeat a string N times
# =============================================

print("=== repeat() ===")
df_rep = df.withColumn("repeated", repeat(lit("ab"), 3))
display(df_rep)
# "ab" × 3 → "ababab"

# =============================================
# 14. OVERLAY — Replace part of string at position
# =============================================

print("=== overlay() ===")
df_overlay = df.withColumn("overlaid", overlay(col("text"), lit("XXX"), lit(3), lit(5)))
display(df_overlay)
# Replaces 5 chars starting at position 3 with "XXX"

# =============================================
# SUMMARY
# =============================================
# | Function          | What it does                         | Example                            |
# |-------------------|--------------------------------------|------------------------------------|
# | lower()           | Lowercase                            | "SPARK" → "spark"                  |
# | upper()           | Uppercase                            | "spark" → "SPARK"                  |
# | initcap()         | Capitalize each word                 | "hello world" → "Hello World"      |
# | trim/ltrim/rtrim  | Remove whitespace                    | "  hi  " → "hi"                    |
# | length()          | String length                        | "spark" → 5                        |
# | concat()          | Join strings                         | "a" + "b" → "ab"                   |
# | concat_ws()       | Join with separator                  | "a", "b" → "a-b"                   |
# | substring()       | Extract part (1-based)               | "spark"(1,3) → "spa"               |
# | split()           | Split into array                     | "a b" → ["a","b"]                  |
# | instr()           | Find position of substring           | "spark","ar" → 3                   |
# | lpad/rpad()       | Pad string                           | lpad("5",3,"0") → "005"            |
# | regexp_replace()  | Replace with regex                   | Remove digits, mask text           |
# | regexp_extract()  | Extract with regex                   | Get email username                 |
# | translate()       | Character-level replace              | "abc","ab","12" → "12c"            |
# | reverse()         | Reverse string                       | "spark" → "kraps"                  |
# | repeat()          | Repeat N times                       | repeat("ab",3) → "ababab"          |
# | overlay()         | Replace at position                  | Insert/replace substring           |
```

###### Date Functions

```python
# =============================================
# DATE FUNCTIONS — Complete Training Demo
# =============================================

from pyspark.sql.functions import (
    col, current_date, current_timestamp, date_add, date_sub,
    datediff, months_between, add_months,
    year, month, dayofmonth, dayofweek, dayofyear,
    date_format, to_date, to_timestamp,
    last_day, next_day, trunc, date_trunc
)

# --- Sample DataFrame ---
df = spark.createDataFrame(
    [
        (1, "Alice", "2026-04-30", "2026-01-01"),
        (2, "Bob", "2025-06-15", "2026-04-30"),
        (3, "Charlie", "2024-12-25", "2026-03-10"),
    ],
    ["id", "name", "start_date", "end_date"]
).withColumn("start_date", col("start_date").cast("date")) \
 .withColumn("end_date", col("end_date").cast("date"))

print("=== Original Data ===")
display(df)

# =============================================
# 1. CURRENT_DATE() & CURRENT_TIMESTAMP()
# =============================================

print("=== current_date() & current_timestamp() ===")
df_now = df.select(
    current_date().alias("today"),
    current_timestamp().alias("now")
)
display(df_now)

# =============================================
# 2. DATE_ADD() — Add days
# =============================================

print("=== date_add() ===")
df_add = df.withColumn("plus_7_days", date_add(col("start_date"), 7))
display(df_add)

# =============================================
# 3. DATE_SUB() — Subtract days
# =============================================

print("=== date_sub() ===")
df_sub = df.withColumn("minus_7_days", date_sub(col("start_date"), 7))
display(df_sub)

# =============================================
# 4. DATEDIFF() — Difference in days between two dates
# =============================================

print("=== datediff() ===")
df_diff = df.withColumn("days_between", datediff(col("end_date"), col("start_date")))
display(df_diff)
# end_date - start_date = number of days

# =============================================
# 5. MONTHS_BETWEEN() — Difference in months
# =============================================

print("=== months_between() ===")
df_months = df.withColumn("months_between", months_between(col("end_date"), col("start_date")))
display(df_months)

# =============================================
# 6. ADD_MONTHS() — Add months to a date
# =============================================

print("=== add_months() ===")
df_add_m = df.withColumn("plus_3_months", add_months(col("start_date"), 3))
display(df_add_m)

# =============================================
# 7. YEAR(), MONTH(), DAYOFMONTH(), DAYOFWEEK(), DAYOFYEAR()
# =============================================

print("=== Extract date parts ===")
df_parts = (
    df
    .withColumn("year", year(col("start_date")))
    .withColumn("month", month(col("start_date")))
    .withColumn("day", dayofmonth(col("start_date")))
    .withColumn("day_of_week", dayofweek(col("start_date")))     # 1=Sunday, 7=Saturday
    .withColumn("day_of_year", dayofyear(col("start_date")))
)
display(df_parts)

# =============================================
# 8. DATE_FORMAT() — Convert date to string
# =============================================

print("=== date_format() ===")
df_fmt = (
    df
    .withColumn("fmt_dd_MM_yyyy", date_format(col("start_date"), "dd-MM-yyyy"))
    .withColumn("fmt_MMM_yyyy", date_format(col("start_date"), "MMM-yyyy"))
    .withColumn("fmt_EEEE", date_format(col("start_date"), "EEEE"))  # Day name
)
display(df_fmt)
# "2026-04-30" → "30-04-2026", "Apr-2026", "Thursday"

# =============================================
# 9. TO_DATE() — Convert string to date
# =============================================

print("=== to_date() ===")
df_str = spark.createDataFrame(
    [(1, "30-04-2026"), (2, "15/06/2025")],
    ["id", "date_str"]
)
df_to_date = (
    df_str
    .withColumn("parsed_date1", to_date(col("date_str"), "dd-MM-yyyy"))
    .withColumn("parsed_date2", to_date(col("date_str"), "dd/MM/yyyy"))
)
display(df_to_date)

# =============================================
# 10. TO_TIMESTAMP() — Convert string to timestamp
# =============================================

print("=== to_timestamp() ===")
df_ts = spark.createDataFrame(
    [(1, "2026-04-30 14:30:00"), (2, "2025-06-15 09:15:30")],
    ["id", "ts_str"]
)
df_to_ts = df_ts.withColumn("timestamp", to_timestamp(col("ts_str"), "yyyy-MM-dd HH:mm:ss"))
display(df_to_ts)

# =============================================
# 11. LAST_DAY() — Last day of the month
# =============================================

print("=== last_day() ===")
df_last = df.withColumn("last_day_of_month", last_day(col("start_date")))
display(df_last)
# "2026-04-30" → "2026-04-30" (April has 30 days)

# =============================================
# 12. NEXT_DAY() — Next occurrence of a weekday
# =============================================

print("=== next_day() ===")
df_next = df.withColumn("next_monday", next_day(col("start_date"), "Mon"))
display(df_next)

# =============================================
# 13. TRUNC() — Truncate date to unit
# =============================================

print("=== trunc() ===")
df_trunc = (
    df
    .withColumn("trunc_month", trunc(col("start_date"), "MM"))    # First day of month
    .withColumn("trunc_year", trunc(col("start_date"), "YY"))     # First day of year
)
display(df_trunc)
# "2026-04-30" → "2026-04-01" (month), "2026-01-01" (year)

# =============================================
# 14. DATE_TRUNC() — Truncate timestamp to unit
# =============================================

print("=== date_trunc() ===")
df_dt = spark.createDataFrame(
    [(1, "2026-04-30 14:35:22")],
    ["id", "ts"]
).withColumn("ts", to_timestamp(col("ts")))

df_date_trunc = (
    df_dt
    .withColumn("trunc_hour", date_trunc("hour", col("ts")))     # 2026-04-30 14:00:00
    .withColumn("trunc_day", date_trunc("day", col("ts")))       # 2026-04-30 00:00:00
    .withColumn("trunc_month", date_trunc("month", col("ts")))   # 2026-04-01 00:00:00
)
display(df_date_trunc)

# =============================================
# SUMMARY
# =============================================
# | Function           | What it does                          | Example                           |
# |--------------------|---------------------------------------|------------------------------------|
# | current_date()     | Today's date                          | → 2026-04-30                      |
# | current_timestamp()| Current date + time                   | → 2026-04-30 14:30:00             |
# | date_add()         | Add N days                            | 04-30 + 7 → 05-07                 |
# | date_sub()         | Subtract N days                       | 04-30 - 7 → 04-23                 |
# | datediff()         | Days between two dates                | 04-30 minus 01-01 → 119           |
# | months_between()   | Months between two dates              | → 3.96                            |
# | add_months()       | Add N months                          | 04-30 + 3 → 07-30                 |
# | year/month/day()   | Extract date parts                    | 2026, 4, 30                       |
# | dayofweek()        | Day of week (1=Sun, 7=Sat)            | → 5 (Thursday)                    |
# | date_format()      | Date → formatted string               | → "30-Apr-2026"                   |
# | to_date()          | String → date                         | "30-04-2026" → date               |
# | to_timestamp()     | String → timestamp                    | "2026-04-30 14:30" → timestamp    |
# | last_day()         | Last day of month                     | → 2026-04-30                      |
# | next_day()         | Next weekday occurrence               | next Monday → 2026-05-04          |
# | trunc()            | Truncate date to month/year           | → 2026-04-01                      |
# | date_trunc()       | Truncate timestamp to hour/day/month  | → 2026-04-30 14:00:00             |
```

###### Handling Nulls

```python
# =============================================
# DROPPING NULL VALUES
# =============================================

# 1. dropna() / dropna(how="any") — Drop rows where ANY column is NULL
df_drop_null = df_delta.dropna()
display(df_drop_null)

# 2. dropna(how="all") — Drop rows where ALL columns are NULL
df_drop_null_all = df_delta.dropna(how="all")
display(df_drop_null_all)

# 3. dropna(subset=["col1", "col2"]) — Drop rows where specific columns are NULL
df_drop_null_subset = df_delta.dropna(subset=["TenantStatus", "TenantMovement"])
display(df_drop_null_subset)

# =============================================
# SUMMARY
# =============================================
# | Method                          | Drops row when                          |
# |---------------------------------|-----------------------------------------|
# | dropna() / dropna(how="any")    | ANY column is NULL                      |
# | dropna(how="all")               | ALL columns are NULL                    |
# | dropna(subset=["col1","col2"])   | Specified columns are NULL (any)        |
# | dropna(how="all", subset=[...]) | ALL specified columns are NULL          |
```

```python
# =============================================
# FILLING NULL VALUES
# =============================================

# 1. fillna() — Fill ALL null values with a single value
df_fill_all = df_delta.fillna("Unknown")
display(df_fill_all)

# 2. fillna() — Fill nulls with different values per column
df_fill_cols = df_delta.fillna({
    "TenantStatus": "Unknown",
    "TenantMovement": "Unclassified",
    "TotalSubscriptionCount": 0
})
display(df_fill_cols)

# 3. fillna() — Fill only specific columns with same value
df_fill_subset = df_delta.fillna("N/A", subset=["TenantStatus", "TenantMovement"])
display(df_fill_subset)

# 4. Using when() + isNull() — Conditional fill (more control)
from pyspark.sql.functions import col, when

df_fill_when = df_delta.withColumn(
    "TenantStatus",
    when(col("TenantStatus").isNull(), "Unknown").otherwise(col("TenantStatus"))
)
display(df_fill_when)

# 5. Using coalesce() — First non-null value from multiple columns
from pyspark.sql.functions import coalesce, lit

df_fill_coalesce = df_delta.withColumn(
    "TenantStatus",
    coalesce(col("TenantStatus"), col("TenantMovement"), lit("Default"))
)
display(df_fill_coalesce)
# Picks first non-null: TenantStatus → TenantMovement → "Default"

# =============================================
# SUMMARY
# =============================================
# | Method                              | What it does                                  |
# |-------------------------------------|-----------------------------------------------|
# | fillna("val")                       | Fill ALL nulls with same value                |
# | fillna({"col1": "v1", "col2": 0})   | Fill nulls per column with different values   |
# | fillna("val", subset=["col1"])      | Fill nulls only in specific columns           |
# | when(isNull(), "val").otherwise()   | Conditional fill with more control            |
# | coalesce(col1, col2, lit("def"))    | First non-null from multiple columns          |
```

###### Split & Indexing

```python
# =============================================
# SPLIT & INDEXING — Complete Demo
# =============================================

from pyspark.sql.functions import col, split, size, element_at, explode

# --- Sample DataFrame ---
df = spark.createDataFrame(
    [
        (1, "alice@gmail.com", "Java,Python,Spark"),
        (2, "bob@outlook.com", "SQL,Python"),
        (3, "charlie@yahoo.co.in", "Scala,Spark,Kafka,Flink"),
        (4, "diana@company.org", "Python"),
    ],
    ["id", "email", "skills"]
)

print("=== Original Data ===")
display(df)

# =============================================
# 1. SPLIT — Split string into array
# =============================================

print("=== split() ===")
df_split = df.withColumn("skills_array", split(col("skills"), ","))
display(df_split)
# "Java,Python,Spark" → ["Java", "Python", "Spark"]

# =============================================
# 2. INDEXING — Access array elements
# =============================================

# Method 1: Using [index] — 0-based
print("=== [index] — 0-based ===")
df_index = (
    df
    .withColumn("skills_array", split(col("skills"), ","))
    .withColumn("first_skill", split(col("skills"), ",")[0])      # First element
    .withColumn("second_skill", split(col("skills"), ",")[1])     # Second (NULL if not exists)
    .withColumn("third_skill", split(col("skills"), ",")[2])      # Third (NULL if not exists)
)
display(df_index)

# Method 2: Using element_at() — 1-based (supports negative index)
print("=== element_at() — 1-based ===")
df_elem = (
    df
    .withColumn("skills_array", split(col("skills"), ","))
    .withColumn("first", element_at(col("skills_array"), 1))      # First element
    .withColumn("last", element_at(col("skills_array"), -1))      # LAST element
)
display(df_elem)

# =============================================
# 3. SIZE — Get array length
# =============================================

print("=== size() ===")
df_size = (
    df
    .withColumn("skills_array", split(col("skills"), ","))
    .withColumn("num_skills", size(split(col("skills"), ",")))
)
display(df_size)

# =============================================
# 4. REAL-WORLD — Split email into user & domain
# =============================================

print("=== Split email ===")
df_email = (
    df
    .withColumn("username", split(col("email"), "@")[0])           # "alice"
    .withColumn("domain", split(col("email"), "@")[1])             # "gmail.com"
    .withColumn("domain_name", split(split(col("email"), "@")[1], "\\.")[0])  # "gmail"
)
display(df_email)

# =============================================
# 5. EXPLODE — One row per array element
# =============================================

print("=== explode() ===")
df_explode = (
    df
    .withColumn("skill", explode(split(col("skills"), ",")))
)
display(df_explode)
# 1 row with "Java,Python,Spark" → 3 rows: "Java", "Python", "Spark"

# =============================================
# SUMMARY
# =============================================
# | Method                    | Index Type | Example                              |
# |---------------------------|-----------|--------------------------------------|
# | split(col, ",")           | —         | "a,b,c" → ["a","b","c"]             |
# | split(col, ",")[0]        | 0-based   | First element                        |
# | split(col, ",")[1]        | 0-based   | Second element (NULL if missing)     |
# | element_at(arr, 1)        | 1-based   | First element                        |
# | element_at(arr, -1)       | 1-based   | LAST element                         |
# | size(arr)                 | —         | Array length                         |
# | explode(arr)              | —         | One row per element                  |
```

###### Collect_list & Collect_set

```python
# =============================================
# COLLECT_LIST & COLLECT_SET — Aggregate into Arrays
# =============================================

from pyspark.sql.functions import col, collect_list, collect_set, size, array_sort

# --- Sample DataFrame ---
df = spark.createDataFrame(
    [
        (1, "Engineering", "Alice", "Python"),
        (2, "Engineering", "Bob", "Java"),
        (3, "Engineering", "Alice", "Spark"),
        (4, "Engineering", "Bob", "Python"),
        (5, "Marketing", "Charlie", "SQL"),
        (6, "Marketing", "Diana", "Python"),
        (7, "Marketing", "Charlie", "Excel"),
        (8, "Marketing", "Charlie", "SQL"),
    ],
    ["id", "dept", "name", "skill"]
)

print("=== Original Data ===")
display(df)

# =============================================
# 1. COLLECT_LIST — Collects all values (keeps duplicates)
# =============================================

print("=== collect_list() — keeps duplicates ===")
df_list = df.groupBy("dept").agg(
    collect_list("skill").alias("all_skills")
)
display(df_list)
# Engineering → ["Python", "Java", "Spark", "Python"]  ← "Python" appears twice

# =============================================
# 2. COLLECT_SET — Collects unique values only
# =============================================

print("=== collect_set() — unique only ===")
df_set = df.groupBy("dept").agg(
    collect_set("skill").alias("unique_skills")
)
display(df_set)
# Engineering → ["Python", "Java", "Spark"]  ← no duplicates

# =============================================
# 3. MULTIPLE collect_list/set in one agg
# =============================================

print("=== Multiple aggregations ===")
df_multi = df.groupBy("dept").agg(
    collect_list("name").alias("all_names"),
    collect_set("name").alias("unique_names"),
    collect_set("skill").alias("unique_skills"),
    size(collect_set("skill")).alias("skill_count")
)
display(df_multi)

# =============================================
# 4. SORTED collect_list
# =============================================

print("=== Sorted collect_list ===")
df_sorted = df.groupBy("dept").agg(
    array_sort(collect_set("skill")).alias("skills_sorted")
)
display(df_sorted)
# ["Excel", "Python", "SQL"] — alphabetically sorted

# =============================================
# 5. GROUP BY person — Collect their skills
# =============================================

print("=== Per-person skills ===")
df_person = df.groupBy("name").agg(
    collect_set("skill").alias("skills"),
    size(collect_set("skill")).alias("num_skills")
)
display(df_person)
# Alice → ["Python", "Spark"], 2
# Charlie → ["SQL", "Excel"], 2

# =============================================
# SUMMARY
# =============================================
# | Function         | Duplicates? | Example                                      |
# |------------------|-------------|----------------------------------------------|
# | collect_list()   | YES keeps   | ["Python","Java","Python"] ← duplicates kept |
# | collect_set()    | NO removes  | ["Python","Java"] ← unique only              |
# | array_sort()     | —           | Sorts the resulting array alphabetically      |
# | size()           | —           | Count of elements in the array               |
#
# CAUTION:
# - collect_list/set brings ALL values to one executor → can OOM on large groups
# - Fine for small-to-medium groups (< 10K values per group)
# - For millions of values per group, consider alternatives

# =============================================
# COLLECT_LIST / COLLECT_SET — When to Use & When NOT
# =============================================

# WHY NOT ON LARGE DATA:
# - All values for a group are pulled to ONE executor's memory
# - A group with millions of rows → OOM (Out of Memory)
# - It's an anti-pattern at scale

# =============================================
# SCALABLE ALTERNATIVES
# =============================================
# | Need                       | Use this (scalable)                              |
# |----------------------------|--------------------------------------------------|
# | Count per group            | groupBy().count()                                |
# | Unique count per group     | countDistinct("col")                             |
# | Comma-separated string     | concat_ws after collect_list on small data only   |
# | Flatten/denormalize        | Window functions or joins                        |
# | Top N per group            | row_number().over(Window)                        |

# =============================================
# WHEN COLLECT_LIST/COLLECT_SET IS SAFE
# =============================================
# - After a groupBy().count() that already reduced data to few rows
# - Small lookup/dimension tables (< 100K rows)
# - Final reporting step on aggregated data (few hundred rows)
# - Collecting into a display/export — not for further processing
#
# RULE OF THUMB: If a group can have more than 10K values,
#                don't use collect_list/collect_set
```

###### Pivot

```python
# =============================================
# PIVOT — Rows to Columns
# =============================================

from pyspark.sql.functions import col, sum, count, avg

# --- Sample DataFrame ---
df = spark.createDataFrame(
    [
        ("Alice", "Q1", 1000),
        ("Alice", "Q2", 1500),
        ("Alice", "Q3", 1200),
        ("Alice", "Q4", 1800),
        ("Bob", "Q1", 2000),
        ("Bob", "Q2", 2500),
        ("Bob", "Q3", 2200),
        ("Bob", "Q4", 3000),
        ("Charlie", "Q1", 800),
        ("Charlie", "Q2", 900),
        ("Charlie", "Q3", 1100),
        ("Charlie", "Q4", 1300),
    ],
    ["name", "quarter", "revenue"]
)

print("=== Original Data (Long Format) ===")
display(df)

# =============================================
# 1. BASIC PIVOT — Rows to Columns
# =============================================

print("=== Basic pivot() ===")
df_pivot = (
    df
    .groupBy("name")
    .pivot("quarter")
    .sum("revenue")
)
display(df_pivot)
# name    | Q1   | Q2   | Q3   | Q4
# Alice   | 1000 | 1500 | 1200 | 1800
# Bob     | 2000 | 2500 | 2200 | 3000
# Charlie | 800  | 900  | 1100 | 1300

# =============================================
# 2. PIVOT with specific values (PERFORMANCE BEST PRACTICE)
# =============================================

print("=== pivot() with specific values — FASTER ===")
df_pivot_fast = (
    df
    .groupBy("name")
    .pivot("quarter", ["Q1", "Q2", "Q3", "Q4"])
    .sum("revenue")
)
display(df_pivot_fast)
# Same result but FASTER — Spark doesn't need to scan data to find distinct values

# =============================================
# 3. PIVOT with different aggregations
# =============================================

print("=== pivot() with avg ===")
df_pivot_avg = (
    df
    .groupBy("name")
    .pivot("quarter")
    .avg("revenue")
)
display(df_pivot_avg)

print("=== pivot() with count ===")
df_pivot_count = (
    df
    .groupBy("name")
    .pivot("quarter")
    .count()
)
display(df_pivot_count)

# =============================================
# 4. PIVOT with multiple aggregations using agg()
# =============================================

print("=== pivot() with multiple aggs ===")
df_pivot_multi = (
    df
    .groupBy("name")
    .pivot("quarter")
    .agg(
        sum("revenue").alias("total"),
        count("revenue").alias("count")
    )
)
display(df_pivot_multi)
# Creates columns: Q1_total, Q1_count, Q2_total, Q2_count, ...

# =============================================
# 5. UNPIVOT — Columns back to Rows (reverse of pivot)
# =============================================

print("=== Unpivot (stack) — Columns to Rows ===")
# Note: PySpark has no native unpivot() before Spark 3.4
# stack() via selectExpr is the standard workaround
df_pivoted = (
    df
    .groupBy("name")
    .pivot("quarter", ["Q1", "Q2", "Q3", "Q4"])
    .sum("revenue")
)

# Using stack() to unpivot
df_unpivot = df_pivoted.selectExpr(
    "name",
    "stack(4, 'Q1', Q1, 'Q2', Q2, 'Q3', Q3, 'Q4', Q4) as (quarter, revenue)"
)
display(df_unpivot)
# Back to original long format

# =============================================
# 6. REAL-WORLD EXAMPLE — Sales by Region & Product
# =============================================

print("=== Real-world: Sales by Region ===")
df_sales = spark.createDataFrame(
    [
        ("North", "Laptop", 500),
        ("North", "Phone", 300),
        ("South", "Laptop", 600),
        ("South", "Phone", 400),
        ("North", "Laptop", 200),
        ("South", "Tablet", 350),
    ],
    ["region", "product", "amount"]
)

df_sales_pivot = (
    df_sales
    .groupBy("region")
    .pivot("product")
    .agg(
        sum("amount").alias("total"),
        count("amount").alias("orders")
    )
)
display(df_sales_pivot)

# =============================================
# SUMMARY
# =============================================
# | Concept                         | Code                                            |
# |---------------------------------|-------------------------------------------------|
# | Basic pivot                     | groupBy("col").pivot("pivot_col").sum("val")    |
# | Pivot with values (FASTER)      | .pivot("col", ["v1","v2","v3"])                 |
# | Pivot with avg/count            | .pivot("col").avg("val")                        |
# | Pivot with multiple aggs        | .pivot("col").agg(sum().alias(), count().alias())|
# | Unpivot (reverse)               | selectExpr("stack(N, ...)")                     |
#
# PERFORMANCE TIP:
# - ALWAYS pass values list to pivot(): .pivot("col", ["v1", "v2"])
# - Without it, Spark runs an extra job to find distinct values
# - On large data this extra scan is expensive
```

###### When-Otherwise

```python
# =============================================
# WHEN-OTHERWISE — Conditional Logic (like IF-ELSE)
# =============================================

from pyspark.sql.functions import col, when, lit

# --- Sample DataFrame ---
df = spark.createDataFrame(
    [
        (1, "Alice", 95, 50000),
        (2, "Bob", 42, 30000),
        (3, "Charlie", 78, 75000),
        (4, "Diana", 15, 120000),
        (5, "Eve", 60, None),
        (6, "Frank", None, 45000),
    ],
    ["id", "name", "score", "salary"]
)

print("=== Original Data ===")
display(df)

# =============================================
# 1. BASIC WHEN-OTHERWISE — Single condition (IF-ELSE)
# =============================================

print("=== Single condition ===")
df_basic = df.withColumn(
    "result",
    when(col("score") >= 50, "Pass")
    .otherwise("Fail")
)
display(df_basic)

# =============================================
# 2. MULTIPLE CONDITIONS — (IF-ELSEIF-ELSE)
# =============================================

print("=== Multiple conditions (IF-ELSEIF-ELSE) ===")
df_multi = df.withColumn(
    "grade",
    when(col("score") >= 90, "A")
    .when(col("score") >= 70, "B")
    .when(col("score") >= 50, "C")
    .when(col("score") >= 30, "D")
    .otherwise("F")
)
display(df_multi)

# =============================================
# 3. HANDLING NULLS in when()
# =============================================

print("=== Handling NULLs ===")
df_null = df.withColumn(
    "score_status",
    when(col("score").isNull(), "No Score")
    .when(col("score") >= 50, "Pass")
    .otherwise("Fail")
)
display(df_null)
# Frank (score=NULL) → "No Score"

# =============================================
# 4. MULTIPLE COLUMNS in condition
# =============================================

print("=== Conditions on multiple columns ===")
df_multi_col = df.withColumn(
    "category",
    when((col("score") >= 80) & (col("salary") >= 70000), "Top Performer")
    .when((col("score") >= 50) & (col("salary") >= 40000), "Average")
    .otherwise("Below Average")
)
display(df_multi_col)

# =============================================
# 5. WHEN without OTHERWISE — Returns NULL
# =============================================

print("=== Without otherwise → NULL ===")
df_no_else = df.withColumn(
    "bonus",
    when(col("score") >= 90, "Yes")
    # No otherwise → all non-matching rows get NULL
)
display(df_no_else)

# =============================================
# 6. WHEN with CALCULATIONS
# =============================================

print("=== Calculations in when() ===")
df_calc = df.withColumn(
    "bonus_amount",
    when(col("score") >= 90, col("salary") * 0.20)
    .when(col("score") >= 70, col("salary") * 0.10)
    .when(col("score") >= 50, col("salary") * 0.05)
    .otherwise(lit(0))
)
display(df_calc)
# Alice (95, 50000) → 50000 * 0.20 = 10000

# =============================================
# 7. WHEN inside FILTER
# =============================================

print("=== when() inside filter ===")
df_filter = df.filter(
    when(col("score").isNotNull(), col("score") >= 50)
    .otherwise(False)
)
display(df_filter)

# =============================================
# 8. WHEN inside AGG
# =============================================

print("=== when() inside aggregation ===")
from pyspark.sql.functions import sum, count

df_agg = df.agg(
    count(when(col("score") >= 50, True)).alias("pass_count"),
    count(when(col("score") < 50, True)).alias("fail_count"),
    sum(when(col("score") >= 50, col("salary")).otherwise(lit(0))).alias("pass_total_salary")
)
display(df_agg)

# =============================================
# SUMMARY
# =============================================
# | Pattern                          | Code                                        |
# |----------------------------------|---------------------------------------------|
# | IF-ELSE                          | when(cond, val).otherwise(val)              |
# | IF-ELSEIF-ELSE                   | when().when().when().otherwise()            |
# | NULL check                       | when(col.isNull(), "default")               |
# | AND condition                    | when((cond1) & (cond2), val)                |
# | OR condition                     | when((cond1) | (cond2), val)                |
# | Without otherwise                | Returns NULL for non-matching rows          |
# | With calculation                 | when(cond, col("x") * 0.10)                |
# | Inside agg()                     | count(when(cond, True))                     |
#
# KEY RULES:
# - Always wrap multiple conditions in parentheses: (cond1) & (cond2)
# - when() is evaluated TOP to BOTTOM — first match wins
# - No otherwise() → NULL for non-matching rows
# - Use .isNull() / .isNotNull() for NULL handling
```

###### Joins

```python
# =============================================
# JOINS — Complete Training Demo
# =============================================

from pyspark.sql.functions import col

# --- Sample DataFrames ---
df_employees = spark.createDataFrame(
    [
        (1, "Alice", 101),
        (2, "Bob", 102),
        (3, "Charlie", 103),
        (4, "Diana", 104),
        (5, "Eve", None),
    ],
    ["emp_id", "name", "dept_id"]
)

df_departments = spark.createDataFrame(
    [
        (101, "Engineering"),
        (102, "Marketing"),
        (103, "Finance"),
        (105, "HR"),
    ],
    ["dept_id", "dept_name"]
)

print("=== Employees ===")
display(df_employees)

print("=== Departments ===")
display(df_departments)

# =============================================
# 1. INNER JOIN — Only matching rows from BOTH sides
# =============================================

print("=== INNER JOIN ===")
df_inner = df_employees.join(df_departments, "dept_id", "inner")
display(df_inner)
# Alice(101), Bob(102), Charlie(103) ✅
# Diana(104) ❌ — no matching dept
# Eve(NULL) ❌ — NULL never matches
# HR(105) ❌ — no matching employee

# =============================================
# 2. LEFT JOIN — All rows from LEFT + matching from RIGHT
# =============================================

print("=== LEFT JOIN ===")
df_left = df_employees.join(df_departments, "dept_id", "left")
display(df_left)
# Alice(101), Bob(102), Charlie(103) — with dept_name ✅
# Diana(104) — dept_name = NULL (no match) ✅
# Eve(NULL) — dept_name = NULL ✅
# HR(105) ❌ — not in left table

# =============================================
# 3. RIGHT JOIN — All rows from RIGHT + matching from LEFT
# =============================================

print("=== RIGHT JOIN ===")
df_right = df_employees.join(df_departments, "dept_id", "right")
display(df_right)
# Alice(101), Bob(102), Charlie(103) — matched ✅
# HR(105) — name = NULL (no matching employee) ✅
# Diana(104), Eve(NULL) ❌ — not in right table

# =============================================
# 4. FULL OUTER JOIN — All rows from BOTH sides
# =============================================

print("=== FULL OUTER JOIN ===")
df_full = df_employees.join(df_departments, "dept_id", "outer")
display(df_full)
# Alice(101), Bob(102), Charlie(103) — matched ✅
# Diana(104) — dept_name = NULL ✅
# Eve(NULL) — dept_name = NULL ✅
# HR(105) — name = NULL ✅
# Everyone is included

# =============================================
# 5. LEFT SEMI JOIN — Only LEFT rows that have a match (no right columns)
# =============================================

print("=== LEFT SEMI JOIN ===")
df_semi = df_employees.join(df_departments, "dept_id", "left_semi")
display(df_semi)
# Alice, Bob, Charlie ✅ (they have matching dept)
# Only employee columns returned — NO dept_name column
# Like: WHERE dept_id IN (SELECT dept_id FROM departments)

# =============================================
# 6. LEFT ANTI JOIN — Only LEFT rows that have NO match
# =============================================

print("=== LEFT ANTI JOIN ===")
df_anti = df_employees.join(df_departments, "dept_id", "left_anti")
display(df_anti)
# Diana(104), Eve(NULL) ✅ — no matching department
# Like: WHERE dept_id NOT IN (SELECT dept_id FROM departments)

# =============================================
# 7. CROSS JOIN — Every row × every row (Cartesian product)
# =============================================

print("=== CROSS JOIN ===")
df_small1 = spark.createDataFrame([(1, "A"), (2, "B")], ["id", "val1"])
df_small2 = spark.createDataFrame([(10, "X"), (20, "Y")], ["id2", "val2"])

df_cross = df_small1.crossJoin(df_small2)
display(df_cross)
# 2 rows × 2 rows = 4 rows
# (1,A,10,X), (1,A,20,Y), (2,B,10,X), (2,B,20,Y)
# CAUTION: 1M × 1M = 1 TRILLION rows → OOM

# =============================================
# 8. JOIN on DIFFERENT column names
# =============================================

print("=== Join on different column names ===")
df_emp2 = spark.createDataFrame(
    [(1, "Alice", 101), (2, "Bob", 102)],
    ["emp_id", "name", "department_id"]
)
df_dept2 = spark.createDataFrame(
    [(101, "Engineering"), (102, "Marketing")],
    ["dept_id", "dept_name"]
)

# Column names differ → use condition
df_diff = df_emp2.join(
    df_dept2,
    df_emp2["department_id"] == df_dept2["dept_id"],
    "inner"
)
display(df_diff)

# =============================================
# 9. JOIN on MULTIPLE columns
# =============================================

print("=== Join on multiple columns ===")
df_orders = spark.createDataFrame(
    [(1, "2026-01-01", 100), (1, "2026-01-02", 200), (2, "2026-01-01", 150)],
    ["cust_id", "order_date", "amount"]
)
df_returns = spark.createDataFrame(
    [(1, "2026-01-01", 50), (2, "2026-01-01", 30)],
    ["cust_id", "order_date", "return_amount"]
)

df_multi_join = df_orders.join(df_returns, ["cust_id", "order_date"], "left")
display(df_multi_join)

# =============================================
# 10. SELF JOIN — Join a table with itself
# =============================================

print("=== Self Join (Employee → Manager) ===")
df_org = spark.createDataFrame(
    [
        (1, "Alice", None),
        (2, "Bob", 1),
        (3, "Charlie", 1),
        (4, "Diana", 2),
    ],
    ["emp_id", "name", "manager_id"]
)

df_self = df_org.alias("e").join(
    df_org.alias("m"),
    col("e.manager_id") == col("m.emp_id"),
    "left"
).select(
    col("e.name").alias("employee"),
    col("m.name").alias("manager")
)
display(df_self)
# Bob → Alice, Charlie → Alice, Diana → Bob, Alice → NULL

# =============================================
# SUMMARY
# =============================================
# | Join Type       | Left Rows | Right Rows | Use Case                      |
# |-----------------|-----------|------------|-------------------------------|
# | inner           | Matched   | Matched    | Only matching records         |
# | left            | ALL       | Matched    | Keep all left, enrich         |
# | right           | Matched   | ALL        | Keep all right, enrich        |
# | outer (full)    | ALL       | ALL        | Keep everything               |
# | left_semi       | Matched   | NONE       | EXISTS / IN filter            |
# | left_anti       | Unmatched | NONE       | NOT EXISTS / NOT IN filter    |
# | cross           | ALL × ALL | ALL × ALL  | Cartesian product (careful!)  |
#
# KEY RULES:
# - Same column name: .join(df2, "col", "inner")
# - Different names:  .join(df2, df1["a"] == df2["b"], "inner")
# - Multiple columns: .join(df2, ["col1", "col2"], "inner")
# - Self join: use .alias("a") and .alias("b")
# - CROSS JOIN: avoid on large data → row explosion
```

###### Window Functions

```python
# =============================================
# WINDOW FUNCTIONS — Complete Training Demo
# =============================================

from pyspark.sql.functions import (
    col, sum, avg, count, min, max,
    row_number, rank, dense_rank, ntile,
    lead, lag,
    first, last,
    percent_rank, cume_dist
)
from pyspark.sql.window import Window

# --- Sample DataFrame ---
df = spark.createDataFrame(
    [
        ("Engineering", "Alice", 95000),
        ("Engineering", "Bob", 90000),
        ("Engineering", "Charlie", 95000),
        ("Engineering", "Diana", 85000),
        ("Marketing", "Eve", 75000),
        ("Marketing", "Frank", 80000),
        ("Marketing", "Grace", 80000),
        ("Finance", "Henry", 70000),
        ("Finance", "Ivy", 72000),
        ("Finance", "Jack", 68000),
    ],
    ["dept", "name", "salary"]
)

print("=== Original Data ===")
display(df)

# =============================================
# DEFINE WINDOWS
# =============================================

# Window partitioned by dept, ordered by salary descending
w_dept = Window.partitionBy("dept").orderBy(col("salary").desc())

# Window partitioned by dept (no order — for aggregations)
w_dept_agg = Window.partitionBy("dept")

# =============================================
# 1. ROW_NUMBER — Unique sequential number per group
# =============================================

print("=== row_number() ===")
df_rn = df.withColumn("row_num", row_number().over(w_dept))
display(df_rn)
# Engineering: Alice=1, Charlie=2, Bob=3, Diana=4
# No ties — always unique (arbitrary order for same salary)

# =============================================
# 2. RANK — Same rank for ties, SKIPS next
# =============================================

print("=== rank() ===")
df_rank = df.withColumn("rank", rank().over(w_dept))
display(df_rank)
# Engineering: Alice=1, Charlie=1, Bob=3, Diana=4
# Two 95000 → both rank 1, next is 3 (skips 2)

# =============================================
# 3. DENSE_RANK — Same rank for ties, NO skip
# =============================================

print("=== dense_rank() ===")
df_dense = df.withColumn("dense_rank", dense_rank().over(w_dept))
display(df_dense)
# Engineering: Alice=1, Charlie=1, Bob=2, Diana=3
# Two 95000 → both rank 1, next is 2 (no skip)

# =============================================
# 4. ROW_NUMBER vs RANK vs DENSE_RANK — Side by Side
# =============================================

print("=== Comparison: row_number vs rank vs dense_rank ===")
df_compare = (
    df
    .withColumn("row_number", row_number().over(w_dept))
    .withColumn("rank", rank().over(w_dept))
    .withColumn("dense_rank", dense_rank().over(w_dept))
)
display(df_compare)
# | salary | row_number | rank | dense_rank |
# | 95000  | 1          | 1    | 1          |
# | 95000  | 2          | 1    | 1          |  ← tie
# | 90000  | 3          | 3    | 2          |  ← rank skips, dense doesn't
# | 85000  | 4          | 4    | 3          |

# =============================================
# 5. NTILE — Divide into N equal buckets
# =============================================

print("=== ntile() ===")
df_ntile = df.withColumn("quartile", ntile(4).over(w_dept))
display(df_ntile)
# Divides each dept into 4 equal groups

# =============================================
# 6. LAG — Previous row's value
# =============================================

print("=== lag() ===")
df_lag = (
    df
    .withColumn("prev_salary", lag("salary", 1).over(w_dept))
    .withColumn("prev_2_salary", lag("salary", 2).over(w_dept))
)
display(df_lag)
# lag(col, 1) → value from 1 row before (NULL for first row)
# lag(col, 2) → value from 2 rows before

# =============================================
# 7. LEAD — Next row's value
# =============================================

print("=== lead() ===")
df_lead = (
    df
    .withColumn("next_salary", lead("salary", 1).over(w_dept))
    .withColumn("salary_diff", col("salary") - lead("salary", 1).over(w_dept))
)
display(df_lead)
# lead(col, 1) → value from 1 row after (NULL for last row)

# =============================================
# 8. WINDOW AGGREGATIONS — sum, avg, count, min, max
# =============================================

print("=== Window aggregations ===")
df_agg = (
    df
    .withColumn("dept_total", sum("salary").over(w_dept_agg))
    .withColumn("dept_avg", avg("salary").over(w_dept_agg))
    .withColumn("dept_count", count("salary").over(w_dept_agg))
    .withColumn("dept_max", max("salary").over(w_dept_agg))
    .withColumn("dept_min", min("salary").over(w_dept_agg))
)
display(df_agg)
# Every row gets the department-level aggregate — no groupBy needed

# =============================================
# 9. RUNNING TOTAL — Cumulative sum
# =============================================

print("=== Running total (cumulative sum) ===")
w_running = Window.partitionBy("dept").orderBy("salary").rowsBetween(Window.unboundedPreceding, Window.currentRow)

df_running = df.withColumn("running_total", sum("salary").over(w_running))
display(df_running)
# Adds up salary cumulatively within each dept

# =============================================
# 10. FIRST & LAST — First/last value in window
# =============================================

print("=== first() & last() ===")
df_fl = (
    df
    .withColumn("highest_paid", first("name").over(w_dept))
    .withColumn("lowest_paid", last("name").over(w_dept))
)
display(df_fl)

# =============================================
# 11. PERCENT_RANK & CUME_DIST
# =============================================

print("=== percent_rank() & cume_dist() ===")
df_pct = (
    df
    .withColumn("percent_rank", percent_rank().over(w_dept))
    .withColumn("cume_dist", cume_dist().over(w_dept))
)
display(df_pct)
# percent_rank: (rank - 1) / (total - 1) → 0.0 to 1.0
# cume_dist: rank / total → fraction of rows ≤ current value

# =============================================
# 12. REAL-WORLD — Top 2 earners per department
# =============================================

print("=== Top 2 per department ===")
df_top2 = (
    df
    .withColumn("rn", row_number().over(w_dept))
    .filter(col("rn") <= 2)
    .drop("rn")
)
display(df_top2)

# =============================================
# 13. REAL-WORLD — Salary vs department average
# =============================================

print("=== Salary vs dept average ===")
df_vs_avg = (
    df
    .withColumn("dept_avg", avg("salary").over(w_dept_agg))
    .withColumn("diff_from_avg", col("salary") - col("dept_avg"))
    .withColumn("above_avg", col("salary") > col("dept_avg"))
)
display(df_vs_avg)

# =============================================
# SUMMARY
# =============================================
# | Function         | What it does                           | Ties          |
# |------------------|----------------------------------------|---------------|
# | row_number()     | Unique number per row                  | No ties       |
# | rank()           | Same rank for ties, SKIPS              | 1,1,3,4       |
# | dense_rank()     | Same rank for ties, NO skip            | 1,1,2,3       |
# | ntile(n)         | Divide into n equal buckets            | —             |
# | lag(col, n)      | Value from n rows BEFORE               | NULL at start |
# | lead(col, n)     | Value from n rows AFTER                | NULL at end   |
# | first() / last() | First/last value in window             | —             |
# | sum/avg/count()  | Aggregation without groupBy            | —             |
# | percent_rank()   | Percentile ranking (0 to 1)            | —             |
# | cume_dist()      | Cumulative distribution (0 to 1)       | —             |
#
# WINDOW DEFINITIONS:
# | Pattern                                     | Use Case                |
# |--------------------------------------------|-----------------------  |
# | Window.partitionBy("col")                   | Aggregate per group     |
# | Window.partitionBy("col").orderBy("col")    | Ranking / lag / lead    |
# | .rowsBetween(unboundedPreceding, currentRow) | Running total           |
#
# KEY INTERVIEW QUESTION:
# "row_number vs rank vs dense_rank?"
# row_number: always unique (1,2,3,4)
# rank: ties get same rank, skips (1,1,3,4)
# dense_rank: ties get same rank, no skip (1,1,2,3)
```

### User Defined Functions

```python
# =============================================
# USER DEFINED FUNCTIONS (UDF) — Complete Demo
# =============================================

from pyspark.sql.functions import col, udf, pandas_udf
from pyspark.sql.types import StringType, IntegerType, FloatType, ArrayType

# --- Sample DataFrame ---
df = spark.createDataFrame(
    [
        (1, "Alice", "alice@gmail.com", 85000),
        (2, "Bob", "bob@outlook.com", 42000),
        (3, "Charlie", "charlie@yahoo.com", 120000),
        (4, "Diana", "diana@company.org", 55000),
        (5, "Eve", None, 95000),
    ],
    ["id", "name", "email", "salary"]
)

print("=== Original Data ===")
display(df)

# =============================================
# 1. BASIC UDF — Using @udf decorator
# =============================================

print("=== Basic UDF with @udf decorator ===")

@udf(returnType=StringType())
def salary_grade(salary):
    if salary is None:
        return "Unknown"
    elif salary >= 100000:
        return "A"
    elif salary >= 70000:
        return "B"
    elif salary >= 50000:
        return "C"
    else:
        return "D"

df_grade = df.withColumn("grade", salary_grade(col("salary")))
display(df_grade)

# =============================================
# 2. UDF using udf() function (register separately)
# =============================================

print("=== UDF using udf() function ===")

def extract_domain(email):
    if email is None:
        return None
    return email.split("@")[1]

extract_domain_udf = udf(extract_domain, StringType())

df_domain = df.withColumn("domain", extract_domain_udf(col("email")))
display(df_domain)

# =============================================
# 3. UDF with MULTIPLE parameters
# =============================================

print("=== UDF with multiple parameters ===")

@udf(returnType=FloatType())
def calculate_bonus(salary, name):
    if salary is None:
        return 0.0
    if name and name.startswith("A"):
        return float(salary * 0.15)  # 15% for names starting with A
    return float(salary * 0.10)      # 10% for others

df_bonus = df.withColumn("bonus", calculate_bonus(col("salary"), col("name")))
display(df_bonus)

# =============================================
# 4. UDF returning ARRAY
# =============================================

print("=== UDF returning Array ===")

@udf(returnType=ArrayType(StringType()))
def name_parts(name):
    if name is None:
        return None
    return [name.lower(), name.upper(), name[::-1]]

df_parts = df.withColumn("name_variations", name_parts(col("name")))
display(df_parts)

# =============================================
# 5. REGISTER UDF for SQL — spark.udf.register()
# =============================================

print("=== Register UDF for Spark SQL ===")

def mask_email(email):
    if email is None:
        return None
    user, domain = email.split("@")
    return user[0] + "***@" + domain

spark.udf.register("mask_email_sql", mask_email, StringType())

df.createOrReplaceTempView("employees")

df_sql = spark.sql("""
    SELECT name, email, mask_email_sql(email) AS masked_email 
    FROM employees
""")
display(df_sql)

# =============================================
# 6. PANDAS UDF (Vectorized UDF) — MUCH FASTER
# =============================================

print("=== Pandas UDF (Vectorized) — 10-100x faster ===")

import pandas as pd

@pandas_udf(IntegerType())
def salary_tax(salary: pd.Series) -> pd.Series:
    return (salary * 0.30).astype(int)

df_tax = df.withColumn("tax", salary_tax(col("salary")))
display(df_tax)

# =============================================
# 7. PANDAS UDF with STRING operations
# =============================================

print("=== Pandas UDF with strings ===")

@pandas_udf(StringType())
def name_upper(name: pd.Series) -> pd.Series:
    return name.str.upper()

df_upper = df.withColumn("name_upper", name_upper(col("name")))
display(df_upper)

# =============================================
# 8. WHY PANDAS UDF > REGULAR UDF
# =============================================

# Regular UDF:
#   - Processes ONE ROW at a time
#   - Data serialized: JVM → Python → JVM (per row)
#   - SLOW on large data
#
# Pandas UDF (Vectorized):
#   - Processes a BATCH of rows (as Pandas Series/DataFrame)
#   - Uses Apache Arrow for efficient transfer
#   - 10-100x FASTER than regular UDF

# =============================================
# 9. WHEN TO USE UDF vs BUILT-IN FUNCTIONS
# =============================================

# RULE: ALWAYS prefer built-in functions over UDFs
#
# BUILT-IN (FAST):
#   df.withColumn("upper", upper(col("name")))
#   df.withColumn("grade", when(col("salary") >= 100000, "A").otherwise("B"))
#
# UDF (SLOW — use only when no built-in exists):
#   - Complex business logic
#   - External API calls
#   - Custom parsing (regex not enough)
#   - ML model inference

# =============================================
# SUMMARY
# =============================================
# | Type              | Speed     | Syntax                              | Use Case              |
# |-------------------|-----------|-------------------------------------|-----------------------|
# | @udf decorator    | Slow      | @udf(StringType())                  | Simple custom logic   |
# | udf() function    | Slow      | udf(func, StringType())             | Reusable UDF          |
# | spark.udf.register| Slow      | spark.udf.register("name", func)    | Use in Spark SQL      |
# | @pandas_udf       | FAST      | @pandas_udf(StringType())           | Vectorized operations |
#
# KEY RULES:
# 1. ALWAYS try built-in functions FIRST (fastest, optimized by Catalyst)
# 2. If UDF is unavoidable, use Pandas UDF (10-100x faster than regular)
# 3. Regular UDF = last resort (row-by-row, slow, blocks Catalyst optimizer)
# 4. Always handle NULLs in your UDF — Spark passes None for NULL values
# 5. UDFs are a BLACK BOX to Spark — no predicate pushdown, no optimization
```
